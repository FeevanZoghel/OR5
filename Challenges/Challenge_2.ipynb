```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_excel('Challenge_1_medium.xlsx')


# Overal een dict van maken
jobs = []

for i in range(len(df)):
    job = {}

    for column in df.columns:
        job[column] = df[column][i]

    jobs.append(job)
```

```python
def calculate_tardiness(jobs):

    volgorde_dagen = []
    tardiness = []

    totaal = 0

    for job in jobs:
        totaal += job['ProcessingTime']
        volgorde_dagen.append(totaal)

    for i in range(len(jobs)):
        due_date = jobs[i]['DueDate']

        if volgorde_dagen[i] > due_date:
            tardiness.append(volgorde_dagen[i] - due_date)
        else:
            tardiness.append(0)

    return sum(tardiness)
```

```python
def best_swap(jobs):

    best_jobs = jobs.copy()
    best_tardiness = calculate_tardiness(jobs)

    for job1 in range(len(jobs)):
        for job2 in range(job1 + 1, len(jobs)):

            # Iedere swap begint vanaf dezelfde huidige oplossing
            new_jobs = jobs.copy()

            # Twee jobs swappen
            new_jobs[job1], new_jobs[job2] = new_jobs[job2], new_jobs[job1]

            # Tardiness van deze swap berekenen
            new_tardiness = calculate_tardiness(new_jobs)

            # Controleren of deze beter is
            if new_tardiness < best_tardiness:
                best_tardiness = new_tardiness
                best_jobs = new_jobs.copy()

    return best_jobs, best_tardiness
```

```python
print("Beste volgorde:")

for job in best_jobs:
    print(job['JobName'], end=" ")

print(f'\nBeste tardiness: {best_tardiness} dagen')
```
