```python
import pandas as pd
import numpy as np
import random
import math as m

random.seed(42)
```

```python
df = pd.read_excel('Challenge_1_medium.xlsx')

#Overal een dict van maken

jobs = []


for i in range(len(df)):
    job = {}

    for column in df.columns:
        job[column] = df[column][i]

    jobs.append(job)


jobname = []
processingtime = []
duedate = []
for i in df['JobName']:
    jobname.append(i)
for i in df['ProcessingTime']:
    processingtime.append(i)
for i in df['DueDate']:
    duedate.append(i)

#greedy methode
dag = 0 
totaltardiness = 0
volgorde = []
best = []
for i in range(len(jobname)):
    k = duedate.index(min(duedate))
    dag+= processingtime[k]
    if dag>duedate[k]:
        totaltardiness+= -(duedate[k]-dag)
    volgorde.append(jobname[k])
    jobname.pop(k)
    duedate.pop(k)
    processingtime.pop(k)

best = volgorde.copy()
best_tardiness = bereken_tardiness(best, jobs)
current = best.copy()

print(best, best_tardiness)
```

```python
def swap(volgorde, i, j):
    nieuwe_volgorde = volgorde.copy()
    nieuwe_volgorde[i], nieuwe_volgorde[j] = nieuwe_volgorde[j], nieuwe_volgorde[i]
    return nieuwe_volgorde


def bereken_tardiness(volgorde, jobs):
    dag = 0
    tardiness = 0

    for letter in volgorde:
        for job in jobs:
            if job['JobName'] == letter:
                dag += job['ProcessingTime']

                if dag > job['DueDate']:
                    tardiness += dag - job['DueDate']

    return tardiness

for i in range(len(current)):
    for j in range(i + 1, len(current)):

        nieuwe_volgorde = swap(current, i, j)

        tardiness = bereken_tardiness(nieuwe_volgorde, jobs)


        if tardiness < best_tardiness:
            best_tardiness = tardiness
            best = nieuwe_volgorde.copy()

print(f'De volgorde \n{best} \nheeft een totale tardiness van {best_tardiness}')
```

```python
current = best.copy()
current_tard = best_tardiness

best = current.copy()

temp = 0.86
t_max = 1000
rondes = 10

def random_swap(current):
    a = random.randint(0, len(current) - 1)

    b = random.randint(0, len(current) - 1)
    while a == b:
        b = random.randint(0, len(current) - 1)

    best1 = swap(current, a, b)
    tard1 = bereken_tardiness(best1, jobs)

    return best1, tard1

for i in range(rondes): 
    for i in range(t_max):

        best1, tard1 = random_swap(current)

        # Nieuwe oplossing vergelijken met CURRENT
        verschil = current_tard - tard1

        kans = m.exp(verschil / temp)

        # Nieuwe oplossing is beter dan current
        if verschil > 0:
            current = best1.copy()
            current_tard = tard1

        else:
            getal = np.random.choice([0, 1], p=[1-kans, kans])

            # Slechtere oplossing toch accepteren
            if getal == 1:
                current = best1.copy()
                current_tard = tard1

        # Is current de beste oplossing die we ooit hebben gezien?
        if current_tard < best_tardiness:
            best = current.copy()
            best_tardiness = current_tard

        # Temperatuur na 1000 iteraties verlagen
        if (i + 1) % 1000 == 0:
            temp = temp * 0.9


print(f'Beste volgorde: {best}')
print(f'Tardiness: {best_tardiness}')
```
