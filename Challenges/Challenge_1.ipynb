```python
import pandas as pd
```

```python
df = pd.read_excel('Challenge_1_tiny.xlsx')


#Overal een lijst van maken

jobs = []


for i in range(len(df)):
    job = {}

    for column in df.columns:
        job[column] = df[column][i]

    jobs.append(job)

print(jobs)

```

```python
jobs.sort(key=lambda job: job['DueDate'])
jobs
```

```python
df = pd.read_excel('Challenge_1_tiny.xlsx')

df = df.sort_values(by = ['DueDate'],ascending = True)
df.head()

completion_time = 0
dagen_verwisselen = []
dag = 0

for i in df['ProcessingTime']:
    dagen_verwisselen.append(dag)
    dag += i
    completion_time = dag

    

print(f'Completion time: {completion_time} dagen')

df['Volgorde_dagen'] = dagen_verwisselen

tardiness = df['Volgorde_dagen'] + df['ProcessingTime'] - df['DueDate']
df['Tardiness'] = tardiness
df.loc[df['Tardiness']<=0, 'Tardiness'] = 0

totaalvertraging = df['Tardiness'].sum()
print(f'De totale vetraging is: {totaalvertraging} dagen')

df.head()
```
